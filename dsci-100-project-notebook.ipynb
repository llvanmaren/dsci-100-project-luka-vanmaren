The bar chart compares the mean number of tournament wins across standard seeds (1–16). No. 1 seeds won approximately 3.5 games on average, while teams seeded 6-16 averaged less than one. This suggests that a lower seed number is associated with more tournament wins and could be a useful predictor. However, averages do not show the variation among teams with the same seed. The special 1983 seed-0 designation was excluded from this comparison.

```r
wins_by_seed <- selected_data |>
  filter(seed > 0) |>
  group_by(seed) |>
  summarize(mean_wins = mean(tourney_wins), .groups = "drop")

ggplot(wins_by_seed, aes(x = factor(seed), y = mean_wins)) +
  geom_col() +
  labs(
    title = "Average tournament wins by seed",
    x = "Tournament seed",
    y = "Mean tournament wins",
    caption = "Seed 0, the 1983 opening-round code, was excluded."
  )
```

The jitter plot compares regular season winning percentage with the number of tournament games won. Teams that won four or more tournament games generally had high regular-season winning percentages, but teams with similar percentages often had different tournament outcomes, including zero wins. This suggests that regular season performance may help predict tournament wins but cannot determine the result on its own. Points were shifted slightly vertically to reveal overlapping observations and the actual outcomes remain whole number counts.

```r
# Spread overlapping points slightly in the vertical direction.
ggplot(selected_data, aes(x = reg_wins_pct, y = tourney_wins)) +
  geom_jitter(width = 0, height = 0.12, alpha = 0.25) +
  scale_y_continuous(breaks = 0:6) +
  labs(
    title = "Tournament wins versus pre-tournament win percentage",
    x = "Pre-tournament win percentage",
    y = "Tournament wins", 
    caption = "Points are shifted vertically to reduce overlap."
  ) + theme(plot.caption = element_text(hjust = 0.5))
```
